# Sortowanie bąbelkowe
Sortowanie bąbelkowe (ang. Bubble Sort) to jeden z najprostszych algorytmów sortowania. Jego działanie polega na wielokrotnym porównywaniu sąsiednich elementów tablicy i zamienianiu ich miejscami, jeśli są w złej kolejności.
Algorytm „wypycha” największy (lub najmniejszy – zależnie od kierunku sortowania) element na koniec tablicy w każdej iteracji – stąd nazwa bąbelkowe (elementy „wypływają” jak bąbelki w wodzie).

Algorytm sortowania bąbelkowego przy porządkowaniu zbioru nieposortowanego
ma klasę czasowej złożoności obliczeniowej równą O(n2). Sortowanie odbywa się
w miejscu.

- Algorytm sortowania bąbelkowego jest uważany za bardzo zły algorytm sortujący. Można go stosować tylko dla niewielkiej liczby elementów w sortowanym zbiorze (do około 5000). Przy większych zbiorach czas sortowania może być zbyt

# Algorytm sortowania bąbelkowego w C++

```python
#include <iostream>
using namespace std;

void bubbleSort(int tab[], int n) {
    for(int i = 0; i < n - 1; i++) {
        for(int j = 0; j < n - i - 1; j++) {
            if(tab[j] > tab[j + 1]) {
                swap(tab[j], tab[j + 1]);
            }
        }
    }
}

int main() {
    int tab[] = {5, 4, 3, 2, 1};
    int n = 5;

    bubbleSort(tab, n);

    for(int i = 0; i < n; i++)
        cout << tab[i] << " ";

    return 0;
}
```
