# Dwukierunkowe sortowanie bąbelkowe (Cocktail Shaker Sort)
Dwukierunkowe sortowanie bąbelkowe jest rozszerzeniem klasycznego sortowania bąbelkowego.

Działanie polega na tym, że w każdym przebiegu algorytm:
1. Przesuwa największy element w prawo (jak klasyczne sortowanie bąbelkowe).
2. Następnie przesuwa najmniejszy element w lewo.

Dzięki temu elementy „wypływają” jednocześnie z obu końców tablicy, co skraca liczbę przebiegów potrzebnych do posortowania tablicy.

# Algorytm sortowania bąbelkowego w C++

```python
#include <iostream>
using namespace std;

void cocktailShakerSort(int tab[], int n) {
    bool zamiana = true;

    while (zamiana) {
        zamiana = false;

        // Przesuwamy w prawo
        for (int i = 0; i < n-1; i++) {
            if (tab[i] > tab[i + 1]) {
                swap(tab[i], tab[i + 1]);
                zamiana = true;
            }
        }

        if (!zamiana) break;

        zamiana = false;

        // Przesuwamy w lewo
        for (int i = n - 1 - 1; i >= 0; i--) {
            if (tab[i] > tab[i + 1]) {
                swap(tab[i], tab[i + 1]);
                zamiana = true;
            }
        }
    }
}

int main() {
    int tab[] = {5, 3, 8, 4, 2};
    int n = 5;

    cocktailShakerSort(tab, n);

    for (int i = 0; i < n; i++)
        cout << tab[i] << " ";

    return 0;
}
```

# Biblioteka chrono

Biblioteka <chrono> w C++ służy do mierzenia czasu i operowania na jednostkach czasu (milisekundy, mikrosekundy itp.).
Najczęściej używamy jej do:
- pomiaru czasu wykonania funkcji lub algorytmu,
- synchronizacji lub opóźnień w programie,
- operacji na czasie rzeczywistym.

Podstawowe elementy:
- high_resolution_clock – zegar o najwyższej dostępnej rozdzielczości, idealny do pomiarów krótkotrwałych operacji.
- duration – przechowuje różnicę między dwoma punktami w czasie.
- duration_cast – pozwala przekształcić wynik pomiaru np. na milisekundy lub mikrosekundy.

### Pomiar czasu prostego bloku kodu:

```python
#include <iostream>
#include <chrono>
using namespace std;
using namespace std::chrono;

int main() {
    auto start = high_resolution_clock::now(); // start pomiaru

    // Przykładowy kod do pomiaru
    for(int i = 0; i < 1000000; i++);

    auto stop = high_resolution_clock::now();  // koniec pomiaru

    auto duration = duration_cast<milliseconds>(stop - start); // czas w ms
    cout << "Czas wykonania: " << duration.count() << " ms" << endl;

    return 0;
}
```
